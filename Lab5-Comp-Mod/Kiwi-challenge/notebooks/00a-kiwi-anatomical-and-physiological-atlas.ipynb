## 2. Macroscopic 3D Organ Architecture: Optical Cross-Sections $\leftrightarrow$ 3-Plane 7T MRI

### 2.1 Botanical Architecture of a Syncarpous Berry
Botanically, a kiwifruit is a **multi-locular syncarpous berry** — meaning that during floral development, **20 to 40 individual seed-bearing leaves (*carpels*)** fuse side-by-side in a ring around a central floral axis:
- **Hexaploid Green `'Hayward'` (*Actinidia chinensis* var. *deliciosa*, $2n = 6x = 174$):** Retains green chlorophyll in its outer pericarp even when ripe and typically forms **`30–36` radial locules** (`33` in our 2005 7T MRI scan!).
- **Diploid / Tetraploid Gold Cultivars (*Actinidia chinensis* var. *chinensis*, $2x/4x$):** Degrade green chlorophyll during ripening (`AcSGR1/2` pathway), unmasking golden carotenoids, and form **fewer radial locules (`~20–28`)** due to their smaller meristem ring circumference (`NB 02` Module 1).
- **Red-Ringed Cultivars (*A. chinensis*, e.g. `'Red Passion'` / `'Hongyang'`):** Synthesize crimson **anthocyanins** specifically inside the **inner pericarp & locular ring**, providing a natural histological stain that visually demarcates the inner seed-bearing zone from both the outer pericarp and central core!

### 2.2 Why Optical Brightness and 7T $T_2$-MRI Brightness Invert!
In a 7 Tesla $T_2$-weighted RARE spin-echo scan (`TR = 18.5 s`, `TE = 63.9 ms`), voxel brightness $S \propto M_0\,e^{-\text{TE}/T_2}$ measures **how freely water protons tumble** over `64 milliseconds` — *not* optical reflectance:
1. **Central Columella (`White optically → Dark in MRI`):** Scattering of visible light off dense, air-interspaced cellulose/pectin cell walls makes the core look chalky white to your eye. Inside the 7T magnet, however, water tightly bound to that dense cell-wall matrix dephases rapidly (short $T_2$), turning the core **dark**!
2. **Locular Mucilage (`Translucent optically → Brightest in MRI`):** Each seed chamber is filled with a watery **pectic/galacturonan hydrogel** (`long T2 ≈ 180 ms`), so at `TE = 63.9 ms` the 33 radial spokes glow like bright wheel spokes!
3. **Mature Seeds (`Dark optically → Jet-Black in MRI`):** Dehydrated, lignified seed coats (`T2 ≈ 14 ms`) lose `>99%` of their transverse magnetization before `TE = 63.9 ms`, punching crisp black holes inside the bright mucilage spokes.

## 4. Microscopic Histology & Cellular Architecture ($\mu\text{m}$ Scale)

Zooming in $100\times$ with light and polarized-light microscopy reveals *why* each tissue zone behaves the way it does in our `Lab 5` biophysical models:

1. **Central Columella (`Central Placenta` — *Guo et al., 2013*, `PMC3797463` Fig. 2):** Composed of small, tightly packed mechanical parenchyma cells with thick primary walls and intercellular matrix. Unlike the watery locules, this core resists radial compression — acting like a stiff elastic rod (`fibrous mediastinum` or `intervertebral annulus`).
2. **Axile Placenta, Septa & Amphicribral Vascular Bundles (`PMC3797463` Figs. 3 & 5):** Where the ~33 radial carpels meet the central core, specialized **amphicribral vascular bundles** (*concentric phloem completely surrounding central water-conducting xylem tracheids!*) branch into every ovule stalk (*funiculus*):
   - **Inward to the fruit (`Sink import`):** Phloem unloads leaf-derived sucrose and water into growing pericarp vacuoles (`NB 01` Stage 2B).
   - **Outward from fertilized seeds (`Paracrine signaling`):** Each fertilized seed synthesizes **auxin (`IAA`) and cytokinins** that diffuse radially (`decay length ℓ ≈ 7.5 mm`, `NB 01` Stage 3B) across the outer pericarp to stimulate local cell expansion!
3. **Outer Pericarp Dimorphic Parenchyma (`PMC3797463` Fig. 4):** A striking two-population tissue where **giant thin-walled sac cells (`500–800 μm` diameter — visible to the naked eye!)** sit embedded in a mesh of smaller dividing cells (`100–200 μm`) irrigated by collateral vascular bundles. When internal hydrostatic **turgor pressure $P$** exceeds the cell-wall yield threshold $Y$, expansin enzymes loosen cellulose microfibrils and these giant cells balloon irreversibly (**Lockhart's viscoplastic growth law**, `NB 01` Stage 2B).
4. **Amyloplast Starch Granules & Lignified Seed *Testa*:**
   - **Why pack starch in `Stage II` (`55–115 DAA`)?** If a growing kiwifruit stored `8%` of its weight directly as free glucose, intracellular osmotic pressure $\Pi = R T \sum c_i$ would skyrocket by `>1.1 MPa`, rupturing cells or halting phloem import! Instead, cells polymerize incoming sugar into **insoluble semicrystalline starch granules inside amyloplasts** (which glow with classic *"Maltese cross"* birefringence under polarized light) — the exact plant counterpart of **human hepatocyte & skeletal-muscle glycogen granules**!
   - **Why do seeds turn jet-black in 7T MRI?** As the embryo matures, the outer seed coat (*testa*) differentiates into thick-walled, honeycomb-pitted **lignified sclereids** and expels free water, dropping $T_2$ from `~95 ms` to `~14 ms` while keeping the dormant embryo alive inside (`Plate 2F`).

## 5. Life Cycle, Dioecious `XY` Sex Dimorphism & the `BBCH` Developmental Clock

Unlike tomatoes or apples (whose flowers carry *both* functional male pollen and a functional female ovary in the same blossom), all wild and commercial *Actinidia* species are **dioecious perennials** — separate **male (`XXXXXY`)** and **female (`XXXXXX`)** climbing vines:

1. **The Two-Gene Y-Chromosome Sex Switch (*Akagi et al., 2018, 2019*):**
   - Both male and female flower buds initiate primordia for stamens *and* a multi-carpellate ovary!
   - On **male (`XY`) vines**, the Y-linked cytokinin response regulator ***Shy Girl* (`SyGI`, GenBank `LC260493.1`)** switches ON at the center of the floral meristem and **represses carpel/ovary expansion**, leaving only a tiny sterile central nub (*pistillode*) surrounded by `>100` fertile yellow pollen-producing anthers (`Plate 3B`), while a second Y-linked gene, ***Friendly Boy* (`FrBy`)**, maintains tapetum/pollen viability.
   - On **female (`XX`) vines**, `SyGI` and `FrBy` are absent: the central syncarpous ovary swells, topped by **`~33` feathery white stigmatic arms** (`Plate 3A` — **one stigma per radial seed chamber!**), while her anthers produce sterile, empty pollen grains!
2. **Pollination Stoichiometry (`Why Orchards Need Bees & Male Pollinizers`):**
   - Because each female flower contains `~1,200–1,500` ovules across its ~33 carpels — and each seed requires **its own pollen grain** growing a pollen tube down one of the ~33 styles — a single `'Hayward'` flower must receive **`1,000–1,400+` viable male pollen grains** within ~5 days of opening (**anthesis, `0 DAA` / `BBCH 65`**) to reach full `85–100 g` commercial size!
   - If wind or honeybees only dust one side of the stigma crown, only the carpels on that side set seeds, producing a **flat-sided asymmetric fruit** (`NB 01` Stage 3B & `NB 02` Module 3).
3. **Standardizing Time Across Seasons (`BBCH` Scale — *Richardson et al., 2011*, `PMC3261216`):**
   - Just as human embryology uses **Carnegie Stages** rather than calendar dates alone, plant physiologists align kiwifruit development to the decimal **`BBCH` phenological scale**: `BBCH 51–59` (bud swell & inflorescence emergence) $\to$ `BBCH 65` (full bloom / anthesis, `0 DAA`) $\to$ `BBCH 71–79` (cell division & exponential fruit expansion) $\to$ `BBCH 81–85` (physiological harvest maturity & seed coat hardening) $\to$ `BBCH 89–90` (eating-ripe climacteric softening & senescence).

## 6. Integrated Physiology (`0–180 DAA`) & the Translational Bridge to Human Medicine

Putting anatomy, histology, and phenology together gives us a complete **four-phase physiological narrative** of the kiwifruit — and reveals a remarkable **translational bridge to human clinical physiology**:

| Developmental Phase | Time (`DAA` / `BBCH`) | Dominant Cellular & Biochemical Processes | 7T MRI Signature (`TE = 63.9 ms`) | Direct Human Biomedical Parallel |
| :--- | :--- | :--- | :--- | :--- |
| **Phase I: Cell Division & Auxin/Cytokinin Drive** | `0–55 DAA` (`BBCH 65–75`) | Fertilized ovules secrete **auxin + cytokinins**; rapid pericarp mitosis; thin-walled giant cells begin turgor expansion; seeds are hydrated & white | Small rapidly growing organ; **hydrated white ovules have long $T_2$** (`<2×` contrast vs. gel!) | **Trophoblast/placental endocrine drive**; hyperplastic organ growth |
| **Phase II: Starch Sink & Seed Lignification** | `55–115 DAA` (`BBCH 75–79`) | Mitosis stops; phloem unloads massive sucrose polymerized into **osmotically silent amyloplast starch** (`up to ~8% FW`); seed *testa* lignifies & dehydrates (`white → brown → black`) | Volume approaches plateau; **seeds turn jet-black** as $T_2$ plunges from `~95 ms` to `~14 ms`! | **Hepatic glycogen storage** (buffering blood osmolarity); tissue fibrosis/calcification |
| **Phase III: Starch $\to$ Sugar Osmotic Switch** | `115–160 DAA` (`BBCH 80–85`, Harvest) | Cool autumn nights trigger **$\beta$-amylase (`BAM`)** cleaving starch into soluble **glucose + fructose + sucrose** (`°Brix` rises `6% → 14%`), sustaining osmotic turgor | Mature **`86 cm³` berry** (`190×` bright mucilage gel vs. black seeds — **our Feb 2005 7T scan!**) | **Glycogenolysis** & **Starling oncotic/osmotic fluid retention** (`ADPKD` cysts / edema) |
| **Phase IV: Climacteric Ethylene Softening** | `160–180+ DAA` (`BBCH 89`, Eating Ripe) | Autocatalytic **ethylene (`AcACS`/`AcACO`)** burst activates polygalacturonase & pectin methylesterase, solubilizing cell-wall pectins (`firmness drops 80 N → <10 N`); **actinidin (`Act d 1`)** peaks | Outer pericarp $T_2$ lengthens & homogenizes as cell-wall water redistributes | **Cervical ripening / ECM remodeling (`MMPs`)**; lysosomal cathepsin proteases (`PDB 2ACT`) |

### The Full-Circle Clinical Punchline: From 7T Fruit MRI to Human Gut MRI!
Why do gastroenterologists prescribe **two green `'Hayward'` kiwifruits twice daily** as an evidence-based, non-pharmacological therapy for chronic constipation (*Wilkinson-Smith et al., 2019*, [PMID: 30706488](https://pubmed.ncbi.nlm.nih.gov/30706488/); *Gearry et al., 2023*, [PMID: 36537785](https://pubmed.ncbi.nlm.nih.gov/36537785/))?
1. The **pectic mucilage hydrogel** that shines brightly at `T2 ≈ 180 ms` inside the kiwifruit's 33 locules has an extraordinary **water-holding & swelling capacity** (`~3×` that of psyllium fiber) that resists small-intestinal digestion, **trapping mobile water inside the human small bowel and ascending colon** — which Nottingham researchers measured non-invasively in healthy volunteers using **abdominal $T_1 / T_2$ MRI**!
2. Simultaneously, kiwifruit's abundant cysteine protease **actinidin (`Act d 1`, `UniProt P00785`, `PDB 2ACT`)** accelerates gastric cleavage of dietary proteins and modulates gut motility receptors!
